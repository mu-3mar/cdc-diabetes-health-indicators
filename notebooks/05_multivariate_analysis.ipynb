# 04 — Multivariate Analysis

- Investigate selected relationships between multiple features and the target.
- Focus on relationships identified as important during Bivariate Analysis.
- Examine whether associations change when considering additional features.
- Document the main multivariate findings.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/processed/diabetes_cleaned.csv')
df.head().T
```

```python
target_col = "Diabetes_binary"
```

## 1. HighBP × Age → Diabetes
- Diabetes Rate by Age and HighBP

```python
rates = (
    df.groupby(["Age", "HighBP"])[target_col]
    .mean()
    .reset_index()
)

rates["Diabetes_Rate"] = rates[target_col] * 100

sns.lineplot(
    data=rates,
    x="Age",
    y="Diabetes_Rate",
    hue="HighBP",
    marker="o"
)

plt.title("Diabetes Rate by Age and HighBP")
plt.xlabel("Age")
plt.ylabel("Diabetes Rate (%)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 2. HighBP × HighChol → Diabetes
- Diabetes Rate by HighBP and HighChol

```python
rates = (
    df.groupby(["HighBP", "HighChol"])[target_col]
    .mean()
    .reset_index()
)

rates["Diabetes_Rate"] = rates[target_col] * 100

sns.barplot(
    data=rates,
    x="HighBP",
    y="Diabetes_Rate",
    hue="HighChol"
)

plt.title("Diabetes Rate by HighBP and HighChol")
plt.xlabel("HighBP")
plt.ylabel("Diabetes Rate (%)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 3. BMI × Age → Diabetes
- BMI Categorization
- Diabetes Rate by Age and BMI Category

```python
df["BMI_Category"] = pd.cut(
    df["BMI"],
    bins=[0, 18.5, 25, 30, float("inf")],
    labels=["Underweight", "Normal", "Overweight", "Obese"],
    right=False
)
```

```python
rates = (
    df.groupby(["Age", "BMI_Category"], observed=True)[target_col]
    .mean()
    .reset_index()
)

rates["Diabetes_Rate"] = rates[target_col] * 100

sns.lineplot(
    data=rates,
    x="Age",
    y="Diabetes_Rate",
    hue="BMI_Category",
    marker="o"
)

plt.title("Diabetes Rate by Age and BMI Category")
plt.xlabel("Age")
plt.ylabel("Diabetes Rate (%)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 4. GenHlth × Age → Diabetes
- Diabetes Rate by Age and General Health

```python
rates = (
    df.groupby(["Age", "GenHlth"])[target_col]
    .mean()
    .reset_index()
)

rates["Diabetes_Rate"] = rates[target_col] * 100

sns.lineplot(
    data=rates,
    x="Age",
    y="Diabetes_Rate",
    hue="GenHlth",
    marker="o"
)

plt.title("Diabetes Rate by Age and General Health")
plt.xlabel("Age")
plt.ylabel("Diabetes Rate (%)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 5. HighBP × BMI → Diabetes
- Diabetes Rate by BMI Category and HighBP

```python
rates = (
    df.groupby(["BMI_Category", "HighBP"], observed=True)[target_col]
    .mean()
    .reset_index()
)

rates["Diabetes_Rate"] = rates[target_col] * 100

sns.barplot(
    data=rates,
    x="BMI_Category",
    y="Diabetes_Rate",
    hue="HighBP"
)

plt.title("Diabetes Rate by BMI Category and HighBP")
plt.xlabel("BMI Category")
plt.ylabel("Diabetes Rate (%)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 6. Multivariate Findings



- `HighBP` and `Age` showed a clear pattern with the target. Diabetes rates were consistently higher among individuals with `HighBP = 1` across age groups, while diabetes rates generally increased with age.

- The combination of `HighBP` and `HighChol` showed a noticeable difference in diabetes rates. The highest rate was observed when both conditions were present.

- `BMI` and `Age` showed a clear pattern with diabetes. Higher BMI categories generally showed higher diabetes rates, particularly across the older age groups.

- `GenHlth` and `Age` showed a strong pattern with the target. Higher `GenHlth` categories were associated with higher diabetes rates, and the rates generally increased across age groups.

- The combination of `BMI` and `HighBP` showed a clear pattern. Diabetes rates increased across BMI categories, and rates were consistently higher among individuals with `HighBP = 1`.

- Overall, the multivariate analysis showed that several features identified during the bivariate analysis continued to display clear patterns when examined together with another relevant feature. These relationships provide additional context for understanding the target and will inform the next stages of the analysis.
