## 9. Feature Understanding

### Target Variable

- **Diabetes_binary**: The target variable indicating whether the respondent has diabetes/prediabetes or not.

### Binary Features

- **HighBP**: Indicates whether the respondent has high blood pressure.
- **HighChol**: Indicates whether the respondent has high cholesterol.
- **CholCheck**: Indicates whether the respondent had their cholesterol checked within the past 5 years.
- **Smoker**: Indicates whether the respondent is a smoker.
- **Stroke**: Indicates whether the respondent has ever had a stroke.
- **HeartDiseaseorAttack**: Indicates whether the respondent has a history of heart disease or heart attack.
- **PhysActivity**: Indicates whether the respondent engages in physical activity.
- **Fruits**: Indicates whether the respondent consumes fruit at least once per day.
- **Veggies**: Indicates whether the respondent consumes vegetables at least once per day.
- **HvyAlcoholConsump**: Indicates whether the respondent is a heavy alcohol consumer.
- **AnyHealthcare**: Indicates whether the respondent has healthcare coverage.
- **NoDocbcCost**: Indicates whether the respondent was unable to see a doctor because of cost.
- **DiffWalk**: Indicates whether the respondent has serious difficulty walking or climbing stairs.
- **Sex**: Binary representation of the respondent's sex.

### Numeric Feature

- **BMI**: Body Mass Index. It is a continuous numerical feature, with values ranging from 12 to 98 in this dataset.

### Ordinal Categorical Features

- **GenHlth**: General health status represented by ordered categories.
- **Age**: Age represented by ordered age categories rather than exact age.
- **Education**: Education level represented by ordered categories.
- **Income**: Income level represented by ordered categories.

### Other Health Features

- **MentHlth**: Number of days during the past 30 days when mental health was not good.
- **PhysHlth**: Number of days during the past 30 days when physical health was not good.

### Initial Observation

Although all features are stored as numerical values (`float64`), they do not all represent continuous numerical variables. Most features are binary, while `Age`, `Education`, `Income`, and `GenHlth` are ordinal categorical variables, and `BMI` is a continuous numerical feature.
